```python
import pandas as pd
import os

df = pd.read_excel('legume X feature.xlsx')
legumes = df.T.iloc[[0,1],:].dropna(axis='columns').reset_index(drop=True).iloc[1].to_list()
fdf = df.iloc[[0,1],:].dropna(axis='columns').T.set_axis(['param', 'desc'], axis='columns').reset_index(drop=True)
```

```python
# prompt = '''For legume {legume}, I am interested in {feature} ({description}). {scale}
# Construct me the shortest valid deep research prompt which:

# - bases the answer only on data from professional or scientific sources
# - keeps the answer under 200 words, factual, and without speculation
# - provides a reference list with source names

# The prompt will be used directly so do not include any additional text.'''


prompt = '''For legume {legume}, I am interested in {feature} ({description}). {scale}
Construct me the a short, clear and concise deep research prompt which:

- bases the answer only on data from professional or scientific sources
- keeps the answer under 200 words, factual, and without speculation
- returns a reference list with source names

The prompt will be used directly so do not include any additional text.'''


for nlegs, leg in enumerate(legumes):
    # folder = f'raw_prompts/{leg}'
    # os.makedirs(folder, exist_ok=True)
    cnt = 0
    for rid, row in fdf.iterrows():
        cnt += 1
        feat = row.iloc[0].strip()
        desc = row.iloc[1].strip()
        if ':' in feat:
            scale = feat[feat.index(':')+1:].strip()
            feature = feat[:feat.index(':')]
        else:
            scale = None
            feature = feat
        scale_text = f'Scale: {scale}' if scale else ''

        # print(feature)
        # print(desc)
        # print(scale_text)
        # print('')

        p = prompt.format(legume=leg, feature=feature, description=desc, scale=scale_text)
        with open(os.path.join('raw_prompts', f'{nlegs+1:02d}-{leg}-{cnt:02d}-{feature}.txt'), 'w') as ofp:
            ofp.write(p)
```
